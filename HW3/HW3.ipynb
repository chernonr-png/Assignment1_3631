### 1. Breadth-First Tree Search (BFTS)

* **แนวคิด:** ขยายโหนดตามระดับความลึกทีละชั้น (FIFO Queue) โดยไม่เก็บความจำโหนดที่เคยเยี่ยมชมแล้ว (Tree Search อาจเกิดการวนลูปได้หากไม่หยุด)

* **ลำดับการขยายโหนด (Expanded Nodes):**
  1. Arad -> ลูก: Sibiu, Timisoara, Zerind
  2. Sibiu -> ลูก: Arad, Fagaras, Oradea, Rimnicu Vilcea
  3. Timisoara -> ลูก: Arad, Lugoj
  4. Zerind -> ลูก: Arad, Oradea
  5. ขยายต่อไปตามลำดับชั้นจนถึงระดับที่มี Vaslui

* **เส้นทางที่พบ (Path):** Arad -> Sibiu -> Fagaras -> Bucharest -> Urziceni -> Vaslui

* **ระยะทางรวม (Path Cost):** 140 + 99 + 211 + 85 + 142 = **677**

### 2. Depth-First Tree Search (DFTS)

* **แนวคิด:** ขยายโหนดลงไปให้ลึกที่สุดก่อน (LIFO Stack) โดยไม่ตรวจจับการวนลูป (Loop Detector)

* **ผลลัพธ์:** เกิด **Infinite Loop** (ติดลูปวนระหว่าง Arad <-> Sibiu สลับกันไปมาไม่สิ้นสุด)

* **เส้นทางที่พบ (Path):** ไม่พบเส้นทาง (No Path Found)

* **ระยะทางรวม (Path Cost):** N/A

### 3. Breadth-First Search (Graph Search)

* **แนวคิด:** สำรวจแบบกว้างทีละระดับชั้นเหมือน BFS แต่เพิ่ม **Explored Set** เพื่อป้องกันการเยี่ยมชมโหนดซ้ำ

* **ลำดับการขยายโหนด (Expanded Nodes):**
  Arad -> Sibiu -> Timisoara -> Zerind -> Fagaras -> Oradea -> Rimnicu Vilcea -> Lugoj -> Bucharest -> Pitesti -> Craiova -> Mehadia -> Giurgiu -> Urziceni -> Drobeta -> Vaslui

* **เส้นทางที่พบ (Path):** Arad -> Sibiu -> Fagaras -> Bucharest -> Urziceni -> Vaslui

* **ระยะทางรวม (Path Cost):** 140 + 99 + 211 + 85 + 142 = **677**

### 4. Depth-First Graph Search (DFGS)

* **แนวคิด:** ขยายโหนดลงไปให้ลึกที่สุดก่อนโดยเก็บจำโหนดที่เคยเยี่ยมชมแล้ว (เรียงขยายตามตัวอักษร)

* **ลำดับการขยายโหนด (Expanded Nodes):**
  Arad -> Sibiu -> Fagaras -> Bucharest -> Urziceni -> Vaslui

* **เส้นทางที่พบ (Path):** Arad -> Sibiu -> Fagaras -> Bucharest -> Urziceni -> Vaslui

* **ระยะทางรวม (Path Cost):** 140 + 99 + 211 + 85 + 142 = **677**

### 5. Best-First Graph Search

* **แนวคิด:** เลือกขยายโหนดถัดไปตามฟังก์ชันประเมิน f(n) ที่น้อยที่สุดใน Priority Queue

* **หมายเหตุ:** 
  * หากกำหนด f(n) = g(n) จะได้ผลลัพธ์เหมือน Uniform Cost Search (UCS)
  * หากกำหนด f(n) = h(n) จะได้ผลลัพธ์เหมือน Greedy Best-First Search

* **ระยะทางรวม (Path Cost):** **645**

### 6. Uniform Cost Search (UCS / Dijkstra)

* **แนวคิด:** ขยายโหนดตามระยะทางสะสม g(n) จากจุดเริ่มต้นที่น้อยที่สุดเสมอ (การันตีเส้นทางสั้นที่สุด)

* **ลำดับการขยายโหนด (Expanded Nodes):**
  Arad (0) -> Zerind (75) -> Timisoara (118) -> Sibiu (140) -> Rimnicu Vilcea (220) -> Fagaras (239) -> Pitesti (317) -> Bucharest (418) -> Craiova (366) -> Urziceni (503) -> Vaslui (645)

* **เส้นทางที่พบ (Path):** Arad -> Sibiu -> Rimnicu Vilcea -> Pitesti -> Bucharest -> Urziceni -> Vaslui

* **ระยะทางรวม (Path Cost):** 140 + 80 + 97 + 101 + 85 + 142 = **645** 

### 7. Depth-Limited Search (DLS)

* **แนวคิด:** จำกัดความลึกสูงสุด Limit (L) เพื่อป้องกัน Infinite Loop ของ DFS

* **ผลการคำนวณ:**
  * หาก L < 5: ไม่พบคำตอบ (Cutoff)
  * หาก L >= 5: พบคำตอบเดียวกับ DFGS

* **เส้นทางที่พบ (Path):** Arad -> Sibiu -> Fagaras -> Bucharest -> Urziceni -> Vaslui

* **ระยะทางรวม (Path Cost):** 140 + 99 + 211 + 85 + 142 = **677**

### 8. Iterative Deepening Search (IDS)

* **แนวคิด:** เพิ่มระดับความลึก L = 0, 1, 2, 3, 4, 5... ไปเรื่อยๆ และรัน DLS ในแต่ละรอบ

* **ผลการคำนวณ:** ค้นพบคำตอบในรอบที่ L = 5

* **เส้นทางที่พบ (Path):** Arad -> Sibiu -> Fagaras -> Bucharest -> Urziceni -> Vaslui

* **ระยะทางรวม (Path Cost):** 140 + 99 + 211 + 85 + 142 = **677**

### 9. Greedy Best-First Search

* **แนวคิด:** เลือกขยายโหนดโดยพิจารณาเฉพาะค่าประมาณระยะทางที่เหลือไปยังจุดหมาย f(n) = h(n)

* **หมายเหตุ:** ต้องใช้ตารางค่า Heuristic h(n) (Straight-Line Distance ไปยัง Vaslui) ของแต่ละเมือง

* **เส้นทางที่พบ (Path):** Arad -> Sibiu -> Fagaras -> Bucharest -> Urziceni -> Vaslui

* **ระยะทางรวม (Path Cost):** N/A (ขึ้นอยู่กับค่า h(n) ไปยัง Vaslui)

### 10. A*-Search

* **แนวคิด:** ใช้ฟังก์ชันประเมิน f(n) = g(n) + h(n) (การันตีเส้นทางสั้นที่สุดหาก h(n) เป็น Admissible)

* **เส้นทางที่พบ (Path):** Arad -> Sibiu -> Rimnicu Vilcea -> Pitesti -> Bucharest -> Urziceni -> Vaslui

* **ระยะทางรวม (Path Cost):** 140 + 80 + 97 + 101 + 85 + 142 = **645**
